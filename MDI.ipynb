```python
from tkinter import *
import pandas

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LinearRegression

data = pandas.read_csv("loan_approval_random_forest.csv")


class Main:

    def __init__(self):

        self.root = Tk()
        self.root.title("MDI")
        self.root.state("zoomed")

        self.mainMenu = Menu(self.root)
        self.root.config(menu=self.mainMenu)

        self.mainMenu.add_command(
            label="Random Forest",
            command=self.randomForest
        )

        self.mainMenu.add_command(
            label="Decision Tree",
            command=self.decisionTree
        )

        self.mainMenu.add_command(
            label="Linear Regression",
            command=self.LinearRegression
        )

        self.mainLabel = Label(
            self.root,
            text="Loan Approval System",
            font=("Arial", 26)
        )

        self.mainLabel.pack(pady=20)

        self.root.mainloop()


    # ================= RANDOM FOREST =================

    def randomForest(self):

        self.root1 = Toplevel()

        self.root1.title("Random Forest")
        self.root1.geometry("500x500")


        X = data[["Age", "CIBIL", "Income", "ExistingLoan"]]

        Y = data["LoanApproved"]


        X_train, X_test, Y_train, Y_test = train_test_split(
            X,
            Y,
            test_size=0.2,
            random_state=42
        )


        self.model = RandomForestClassifier(
            n_estimators=5,
            random_state=42
        )

        self.model.fit(X_train, Y_train)


        print("Random Forest Training Score:",
              self.model.score(X_train, Y_train))

        print("Random Forest Testing Score:",
              self.model.score(X_test, Y_test))


        self.lb1 = Label(self.root1, text="Customer")
        self.lb2 = Label(self.root1, text="Age")
        self.lb3 = Label(self.root1, text="CIBIL")
        self.lb4 = Label(self.root1, text="Income")
        self.lb5 = Label(self.root1, text="Existing Loan")
        self.lb6 = Label(self.root1, text="Approval")


        self.txt1 = Entry(self.root1)
        self.txt2 = Entry(self.root1)
        self.txt3 = Entry(self.root1)
        self.txt4 = Entry(self.root1)
        self.txt5 = Entry(self.root1)
        self.txt6 = Entry(self.root1)


        self.lb1.grid(row=0, column=0)
        self.txt1.grid(row=0, column=1)

        self.lb2.grid(row=1, column=0)
        self.txt2.grid(row=1, column=1)

        self.lb3.grid(row=2, column=0)
        self.txt3.grid(row=2, column=1)

        self.lb4.grid(row=3, column=0)
        self.txt4.grid(row=3, column=1)

        self.lb5.grid(row=4, column=0)
        self.txt5.grid(row=4, column=1)

        self.lb6.grid(row=5, column=0)
        self.txt6.grid(row=5, column=1)


        self.bt1 = Button(
            self.root1,
            text="Predict",
            command=self.predictRandomForest
        )

        self.bt1.grid(row=6, column=0)


    def predictRandomForest(self):

        age = int(self.txt2.get())
        cibil = int(self.txt3.get())
        income = float(self.txt4.get())
        existingLoan = float(self.txt5.get())


        prediction = self.model.predict([
            [age, cibil, income, existingLoan]
        ])


        self.txt6.delete(0, END)

        self.txt6.insert(0, prediction[0])


    # ================= DECISION TREE =================

    def decisionTree(self):

        self.root2 = Toplevel()

        self.root2.title("Decision Tree")
        self.root2.geometry("500x500")


        X = data[["Age", "CIBIL", "Income", "ExistingLoan"]]

        Y = data["LoanApproved"]


        X_train, X_test, Y_train, Y_test = train_test_split(
            X,
            Y,
            test_size=0.2,
            random_state=42
        )


        self.treeModel = DecisionTreeClassifier(
            random_state=42
        )

        self.treeModel.fit(X_train, Y_train)


        print("Decision Tree Training Score:",
              self.treeModel.score(X_train, Y_train))

        print("Decision Tree Testing Score:",
              self.treeModel.score(X_test, Y_test))


        self.lb7 = Label(self.root2, text="Customer")
        self.lb8 = Label(self.root2, text="Age")
        self.lb9 = Label(self.root2, text="CIBIL")
        self.lb10 = Label(self.root2, text="Income")
        self.lb11 = Label(self.root2, text="Existing Loan")
        self.lb12 = Label(self.root2, text="Approval")


        self.txt7 = Entry(self.root2)
        self.txt8 = Entry(self.root2)
        self.txt9 = Entry(self.root2)
        self.txt10 = Entry(self.root2)
        self.txt11 = Entry(self.root2)
        self.txt12 = Entry(self.root2)


        self.lb7.grid(row=0, column=0)
        self.txt7.grid(row=0, column=1)

        self.lb8.grid(row=1, column=0)
        self.txt8.grid(row=1, column=1)

        self.lb9.grid(row=2, column=0)
        self.txt9.grid(row=2, column=1)

        self.lb10.grid(row=3, column=0)
        self.txt10.grid(row=3, column=1)

        self.lb11.grid(row=4, column=0)
        self.txt11.grid(row=4, column=1)

        self.lb12.grid(row=5, column=0)
        self.txt12.grid(row=5, column=1)


        self.bt2 = Button(
            self.root2,
            text="Predict",
            command=self.predictDecisionTree
        )

        self.bt2.grid(row=6, column=0)


    def predictDecisionTree(self):

        age = int(self.txt8.get())
        cibil = int(self.txt9.get())
        income = float(self.txt10.get())
        existingLoan = float(self.txt11.get())


        prediction = self.treeModel.predict([
            [age, cibil, income, existingLoan]
        ])


        self.txt12.delete(0, END)

        self.txt12.insert(0, prediction[0])


    # ================= LINEAR REGRESSION =================

    def LinearRegression(self):

        self.root3 = Toplevel()

        self.root3.title("Linear Regression")
        self.root3.geometry("500x500")


        X = data[["Age", "CIBIL", "Income", "ExistingLoan"]]

        Y = data["LoanApproved"].map({
            "Yes": 1,
            "No": 0
        })


        X_train, X_test, Y_train, Y_test = train_test_split(
            X,
            Y,
            test_size=0.2,
            random_state=42
        )


        self.linearModel = LinearRegression()

        self.linearModel.fit(X_train, Y_train)


        print("Linear Regression Training Score:",
              self.linearModel.score(X_train, Y_train))

        print("Linear Regression Testing Score:",
              self.linearModel.score(X_test, Y_test))


        self.lb13 = Label(self.root3, text="Customer")
        self.lb14 = Label(self.root3, text="Age")
        self.lb15 = Label(self.root3, text="CIBIL")
        self.lb16 = Label(self.root3, text="Income")
        self.lb17 = Label(self.root3, text="Existing Loan")
        self.lb18 = Label(self.root3, text="Approval")


        self.txt13 = Entry(self.root3)
        self.txt14 = Entry(self.root3)
        self.txt15 = Entry(self.root3)
        self.txt16 = Entry(self.root3)
        self.txt17 = Entry(self.root3)
        self.txt18 = Entry(self.root3)


        self.lb13.grid(row=0, column=0)
        self.txt13.grid(row=0, column=1)

        self.lb14.grid(row=1, column=0)
        self.txt14.grid(row=1, column=1)

        self.lb15.grid(row=2, column=0)
        self.txt15.grid(row=2, column=1)

        self.lb16.grid(row=3, column=0)
        self.txt16.grid(row=3, column=1)

        self.lb17.grid(row=4, column=0)
        self.txt17.grid(row=4, column=1)

        self.lb18.grid(row=5, column=0)
        self.txt18.grid(row=5, column=1)


        self.bt3 = Button(
            self.root3,
            text="Predict",
            command=self.predictLinear
        )

        self.bt3.grid(row=6, column=0)


    def predictLinear(self):

        age = int(self.txt14.get())
        cibil = int(self.txt15.get())
        income = float(self.txt16.get())
        existingLoan = float(self.txt17.get())


        prediction = self.linearModel.predict([
            [age, cibil, income, existingLoan]
        ])


        if prediction[0] >= 0.5:
            result = "Yes"
        else:
            result = "No"


        self.txt18.delete(0, END)

        self.txt18.insert(0, result)


obj = Main()
```
